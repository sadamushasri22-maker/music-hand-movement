```python
import cv2
import mediapipe as mp
import pygame
import time

pygame.mixer.init()

sounds = {
    "C": pygame.mixer.Sound("sounds/C.wav"),
    "D": pygame.mixer.Sound("sounds/D.wav"),
    "E": pygame.mixer.Sound("sounds/E.wav"),
    "F": pygame.mixer.Sound("sounds/F.wav"),
    "G": pygame.mixer.Sound("sounds/G.wav"),
    "A": pygame.mixer.Sound("sounds/A.wav"),
    "B": pygame.mixer.Sound("sounds/B.wav")
}

keys = ["C", "D", "E", "F", "G", "A", "B"]

mp_hands = mp.solutions.hands
hands = mp_hands.Hands(max_num_hands=1)
mp_draw = mp.solutions.drawing_utils

cap = cv2.VideoCapture(0, cv2.CAP_DSHOW)  # IMPORTANT FIX FOR WINDOWS

cv2.namedWindow("Piano", cv2.WINDOW_NORMAL)

last_play_time = 0
cooldown = 0.35

while True:
    success, img = cap.read()
    if not success:
        print("Camera not working")
        break

    img = cv2.flip(img, 1)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    results = hands.process(img_rgb)

    h, w, _ = img.shape
    key_width = w // 7
    key_height = 120

    for i, key in enumerate(keys):
        x1 = i * key_width
        x2 = (i + 1) * key_width

        cv2.rectangle(img, (x1, h - key_height), (x2, h), (255, 255, 255), -1)
        cv2.rectangle(img, (x1, h - key_height), (x2, h), (0, 0, 0), 2)

        cv2.putText(img, key, (x1 + 25, h - 40),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 0), 2)

    if results.multi_hand_landmarks:
        for handLms in results.multi_hand_landmarks:
            mp_draw.draw_landmarks(img, handLms, mp_hands.HAND_CONNECTIONS)

            x = int(handLms.landmark[8].x * w)
            y = int(handLms.landmark[8].y * h)

            cv2.circle(img, (x, y), 10, (0, 0, 255), -1)

            if y > h - key_height:
                key_index = x // key_width

                if 0 <= key_index < 7:
                    current_time = time.time()

                    if current_time - last_play_time > cooldown:
                        note = keys[key_index]
                        sounds[note].play()
                        last_play_time = current_time

    cv2.imshow("Piano", img)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        pygame.mixer.stop()
        break

cap.release()
cv2.destroyAllWindows()
pygame.quit()
```
